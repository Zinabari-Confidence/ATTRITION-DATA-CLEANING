**ADD A COLUMN AGE AND GET AGE FROM DATE OF BIRTH COLUMN**

```python
# Calculate age and add it as a new column
current_date = pd.to_datetime('today')
HR['Age'] = current_date.year - HR['BirthDate'].dt.year - ((current_date.month < HR['BirthDate'].dt.month)
 | ((current_date.month == HR['BirthDate'].dt.month) & (current_date.day < HR['BirthDate'].dt.day))).astype(int)
```

create a conditional column to to meet a particualar condition in the dataframe

```python
# Using apply() and a custom function
def performance_category(emp_performance_ratings):
    if emp_performance_ratings == 1:
        return 'Poor'
    elif emp_performance_ratings == 2:
        return 'Fair'
    elif emp_performance_ratings == 3:
        return 'Good'
    elif emp_performance_ratings == 4:
        return 'Very Good'
    elif emp_performance_ratings == 5:
        return 'Excellent'
    else:
        return 'Not Rated'

# Create a new column for performance categories
HR['Performance Rating'] = HR['emp_performance_ratings'].apply(performance_category)
```
